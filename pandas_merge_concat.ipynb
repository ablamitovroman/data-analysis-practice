```python
import pandas as pd

# 1. Linke Tabelle (Kunden)
data_kunden = {
    "Kunden_ID": [101, 102, 103, 104],
    "Name": ["Anna", "Bernd", "Clara", "Dirk"],
}
df_kunden = pd.DataFrame(data_kunden)

# 2. Rechte Tabelle (Bestellungen)
data_bestellungen = {
    "User_Code": [101, 102, 101, 105],
    "Umsatz": [150.0, 200.0, 50.0, 300.0],
}
df_bestellungen = pd.DataFrame(data_bestellungen)
#print(f"{df_kunden}\n{df_bestellungen}")
# DEIN CODE HIER:
# Verbinde df_kunden und df_bestellungen mit how="left" über "Kunden_ID"
# df_merged = pd.merge(
#     df_kunden,
#     df_bestellungen,
#     left_on="Kunden_ID",
#     right_on="User_Code",
#     how="left"
#     ).drop(columns=['User_Code']
#            ).fillna({"Umsatz": "KEIN"})
# print(df_merged)
# Gib das Ergebnis aus
```

test unterschiedlichen methods für how=

```python
how_methods = ["left", "right", "inner", "outer"]

for method in how_methods:
  df_merged = pd.merge(
      df_kunden,
      df_bestellungen,
      left_on="Kunden_ID",
      right_on="User_Code",
      how=method
  )
  print(f"method:{method}\n{df_merged}\n")
```
