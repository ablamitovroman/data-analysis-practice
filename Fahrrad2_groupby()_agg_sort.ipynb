```python
import pandas as pd

daten = {
    "kategorie": ["E-Bike", "Citybike", "Zubehör", "E-Bike", "Zubehör"],
    "menge": [2, 1, 4, 1, 10],
    "einzelpreis": [1500.0, 300.0, 20.0, 1200.0, 15.0]
}

verkauf = pd.DataFrame(daten)
verkauf["gesamtpreis"] = verkauf["menge"] * verkauf["einzelpreis"]

umsatz_pro_kategorie = verkauf.groupby("kategorie")['gesamtpreis'].sum()
#print(f"\n umsatz_pro_kategorie {umsatz_pro_kategorie}")

menge_pro_kategorie = verkauf.groupby("kategorie")['menge'].sum()
#print(f"\n menge_pro_kategorie {menge_pro_kategorie}")

#print(f"\n verkauf {verkauf}")

#neue_tabelle = verkauf.groupby("kategorie")["gesamtpreis"].agg(["sum", "mean", "count"])

neue_tabelle = verkauf.groupby("kategorie").agg(
    gesamtsumme = ("gesamtpreis", "sum"),
    durchschnitt_preis = ("gesamtpreis", "mean"),
    verkauf_anzahl = ("gesamtpreis", "count"),
    gesamt_menge = ("menge", "sum")
)
#print(neue_tabelle)

sort_tabelle = neue_tabelle.sort_values(by=["verkauf_anzahl", "gesamtsumme"], ascending=False)
#print(sort_tabelle)
sort_gross_umsatz = sort_tabelle[sort_tabelle["gesamtsumme"] > 1000]
print(sort_gross_umsatz)

#umsatz_gross = verkauf[verkauf["gesamtpreis"] > 1000]
#print(f"\n umsatz_gross {umsatz_gross}")

#neue_tabelle = umsatz_gross.groupby("kategorie").agg(
#    gesamtsumme = ("gesamtpreis", "sum"),
#    durchschnitt_preis = ("gesamtpreis", "mean"),
#    verkauf_anzahl = ("gesamtpreis", "count"),
#    gesamt_menge = ("menge", "sum")
#)
#print(f"\n neue_tabelle {neue_tabelle}")
```
