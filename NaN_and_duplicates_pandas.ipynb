Willkommen im neuen **Block 6: Datenbereinigung (Data Cleaning) – Umgang mit fehlenden Werten (`NaN`)**.

In der Praxis fehlen Daten sehr häufig (z.B. wenn Kunden Felder nicht ausfüllen oder Sensoren ausfallen). В Pandas werden fehlende Werte als `NaN` (*Not a Number*) oder `None` dargestellt.

---

### Die 4 wichtigsten Werkzeuge in Pandas

1. **Suchen:** `.isna()` oder `.isnull()`
* Zeigt an, wo Daten fehlen (gibt `True` / `False` zurück).
* `df.isna().sum()` zeigt die **Anzahl der fehlenden Werte** pro Spalte.


2. **Entfernen:** `.dropna()`
* Löscht Zeilen oder Spalten, die `NaN` enthalten.


3. **Ersetzen / Auffüllen:** `.fillna()`
* Ersetzt `NaN` durch einen festen Wert (z.B. `0`, `"Unbekannt"`) oder durch den Mittelwert/Median.


4. **Klassiker:** Fehlende Werte mit dem Mittelwert auffüllen:

```
df["Betrag"] = df["Betrag"].fillna(df["Betrag"].mean())
```

### Übungsaufgabe: Spickzettel "Datenbereinigung"

Hier ist ein DataFrame mit fehlenden Werten:

```
import numpy as np
import pandas as pd

data = {
    "Kunde": ["Anna", "Bernd", "Clara", "Dirk"],
    "Region": ["Nord", np.nan, "Süd", "Nord"],
    "Umsatz": [150.0, np.nan, 300.0, 250.0],
}
df_kunden = pd.DataFrame(data)
print(df_kunden)

df_fehlen_daten = df_kunden.isna().sum()
print(df_fehlen_daten)

df_bereinigt = df_kunden.dropna(subset=["Umsatz"])

df_kunden['Region'] = df_kunden['Region'].fillna("Unbekannt")
df_kunden['Umsatz'] = df_kunden['Umsatz'].fillna(round(df_kunden['Umsatz'].mean()))
print(df_kunden)
print(df_bereinigt)
```

```
import numpy as np
import pandas as pd

data = {
    "Kunden_ID": [101, 102, 103, 104, 105],
    "Stadt": ["Berlin", np.nan, "Hamburg", "München", np.nan],
    "Einkaufswert": [120.0, 50.0, np.nan, 210.0, np.nan],
}
df_kunden = pd.DataFrame(data)
print(f"{df_kunden}\n")
```

fillna() anwenden: Ersetze in der Spalte "Stadt" alle fehlenden Werte (NaN) durch den Text "Unbekannt".

dropna() anwenden: Lösche alle Zeilen, bei denen in der Spalte "Einkaufswert" ein NaN steht (nutze dafür das Argument subset=["Einkaufswert"]).

Ergebnis ausgeben: Gib den bereinigten DataFrame mit print() aus.

```
df_kunden['Stadt'] = df_kunden['Stadt'].fillna("Unbekannt")
df_bereinigt = df_kunden.dropna(subset=["Einkaufswert"])
print(df_bereinigt)
```

```
import pandas as pd

data = {
    "Kunden_ID": [101, 102, 101, 103, 102],
    "Name": ["Anna", "Bernd", "Anna", "Clara", "Bernd"],
    "Umsatz": [150.0, 200.0, 150.0, 300.0, 200.0],
}
df_kunden = pd.DataFrame(data)
print(f"{df_kunden}\n")
df_dupl = df_kunden.duplicated().sum()
print(f"Duplikatten = {df_dupl}\n")
df_unique = df_kunden.drop_duplicates(subset=["Name"])
print(df_unique)
```
