```python
import pandas as pd

daten = {
    "kategorie": ["E-Bike", "Citybike", "Zubehör", "E-Bike", "Zubehör"],
    "menge": [2, 1, 4, 1, 10],
    "einzelpreis": [1500.0, 300.0, 20.0, 1200.0, 15.0]
}

verkauf = pd.DataFrame(daten)
#print(verkauf)

verkauf["gesamtpreis"] = verkauf["menge"] * verkauf["einzelpreis"]
print(verkauf)

total_umsatz = verkauf['gesamtpreis'].sum()
print(f"\n Total Umsatz = {total_umsatz} €")

#viele_artikel = verkauf[verkauf['menge'] > 1]

#total_umsatz_viel = viele_artikel["gesamtpreis"].sum()
total_umsatz_viel = verkauf[verkauf['menge'] > 1]['gesamtpreis'].sum()
print(f"\n Total Umsatz Artikel mit mehr als 1 Stück {total_umsatz_viel} €")
```
