### Подготовка: Создай тестовые файлы

Запусти этот блок кода в Colab. Он создаст **два файла** на диске: один CSV и один Excel.

```
import pandas as pd

# Данные для CSV-файла (Отдел 1)
df_csv = pd.DataFrame([
    {"Mitarbeiter": "Anna", "Umsatz": 3000.0},
    {"Mitarbeiter": "Stefan", "Umsatz": 1500.0},
    {"Mitarbeiter": "Anna", "Umsatz": 4000.0}
])
df_csv.to_csv("vertrieb_nord.csv", index=False)

# Данные для Excel-файла (Отдел 2)
df_excel = pd.DataFrame([
    {"Mitarbeiter": "Markus", "Umsatz": 2000.0},
    {"Mitarbeiter": "Stefan", "Umsatz": 4500.0},
    {"Mitarbeiter": "Markus", "Umsatz": 1000.0}
])
df_excel.to_excel("vertrieb_sued.xlsx", index=False)

print("Файлы 'vertrieb_nord.csv' и 'vertrieb_sued.xlsx' успешно созданы!")
```

### Твоя задача: Класс `MultiDateiAnalyse`

Напиши класс, который умеет принимать **список путей к файлам**, читать их (неважно, CSV это или Excel), объединять все данные в одну общую таблицу и проводить анализ.

#### Что должен делать класс:

1. **Конструктор `__init__(self, datei_liste)**`:
* Принимает список имён/путей к файлам (например, `["vertrieb_nord.csv", "vertrieb_sued.xlsx"]`).
* Сохраняет этот список в `self.datei_liste`.


2. **Метод `daten_laden(self)**`:
* Создаёт пустой список для хранения прочитанных таблиц.
* Проходит циклом `for datei in self.datei_liste:` по всем файлам.
* Проверяет расширение файла:
* Если файл заканчивается на `.csv` — читает его через `pd.read_csv(datei)`.
* Если файл заканчивается на `.xlsx` — читает его через `pd.read_excel(datei)`.


* Объединяет все прочитанные таблицы в один общий DataFrame с помощью функции **`pd.concat(список_таблиц, ignore_index=True)`**.
* Сохраняет итоговый объединённый DataFrame в `self.df_gesamt`.


3. **Метод `analysieren(self)**`:
* Берет объединённую таблицу `self.df_gesamt`.
* Группирует данные по сотрудникам (`Mitarbeiter`, не забудь `as_index=False`).
* Считает:
* **`Gesamtumsatz`** — сумма по столбцу `Umsatz`.
* **`AnzahlDeals`** — количество сделок (`count`).


* Сортирует результат по `Gesamtumsatz` по убыванию.
* Сохраняет итоговый отчёт в `gesamter_bericht.xlsx` (без индексов).
* Возвращает итоговую таблицу через `return`.



---

### Как запустить и проверить:

# Neuer Abschnitt

```
import pandas as pd

class MultiDateiAnalyse:
    def __init__(self, datei_liste):
        self.datei_liste = datei_liste

    def daten_laden(self):
        tabellen = []
        for f in self.datei_liste:
              if f.endswith(".csv"):
                tabellen.append(pd.read_csv(f))
              elif f.endswith(".xlsx"):
                tabellen.append(pd.read_excel(f))
              else:
                print(f"Файл {f} не поддерживается")
        if tabellen:
          self.df_gesamt = pd.concat(tabellen, ignore_index=True)
          print (f"\n Die Daten sind zusammen geführt \n\n {self.df_gesamt}\n")
        else:
          print("Файлы не загружены")

    def analysieren(self):
        ergebnis = (
            self.df_gesamt
              .groupby("Mitarbeiter", as_index=False)
              .agg(
                  Gesamtumsatz = ("Umsatz", "sum"),
                  Anzahldeals = ("Umsatz", "count"))
              .sort_values(by="Gesamtumsatz", ascending=False)
        )

        return ergebnis
```

```
dateien = ["vertrieb_nord.csv", "vertrieb_sued.xlsx"]

analyse = MultiDateiAnalyse(dateien)
analyse.daten_laden()          # Загружаем и объединяем файлы
ergebnis = analyse.analysieren() # Анализируем

print(ergebnis)
```
