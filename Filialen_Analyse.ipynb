### Задача: «Filialen-Analyse» (Анализ продаж по филиалам)

Представь, что тебе дали данные по продажам из трех филиалов (`München`, `Berlin`, `Hamburg`).

#### Что должен уметь твой скрипт:

1. **Класс `VerkaufsDaten` (ООП)**:
* Напиши класс, который при создании принимает список словарей с исходными данными (`Filiale`, `Produkt`, `Menge`, `Einzelpreis`).
* Внутри класса сделай метод, который превращает этот список в `pd.DataFrame` и сразу добавляет новую колонку с общей суммой заказа (`Gesamtsumme` = `Menge` × `Einzelpreis`).


2. **Агрегация данных (Pandas)**:
* Сгруппируй данные по филиалам (`Filiale`).
* Посчитай для каждого филиала: **суммарную выручку** и **средний чек**.
* *Условие:* Сделай так, чтобы название филиала (`Filiale`) осталось обычной колонкой таблицы, а не ушло в индекс (используй `as_index=False` или `.reset_index()`).


3. **Фильтрация и Экспорт**:
* Оставим в отчёте только те филиалы, у которых суммарная выручка **больше 700 €**.
* Сохрани итоговую таблицу в файлы `filialen_bericht.csv` и `filialen_bericht.xlsx` (без системного индекса строк).



---

### Исходные данные для твоей проверки:

```python
import pandas as pd

rohdaten = [
    {"Filiale": "München", "Produkt": "E-Bike", "Menge": 2, "Einzelpreis": 1200.0},
    {"Filiale": "Berlin", "Produkt": "Helm", "Menge": 5, "Einzelpreis": 50.0},
    {"Filiale": "München", "Produkt": "Schloss", "Menge": 3, "Einzelpreis": 30.0},
    {"Filiale": "Hamburg", "Produkt": "Citybike", "Menge": 1, "Einzelpreis": 600.0},
    {"Filiale": "Berlin", "Produkt": "E-Bike", "Menge": 1, "Einzelpreis": 1100.0},
    {"Filiale": "Hamburg", "Produkt": "Klingel", "Menge": 2, "Einzelpreis": 15.0}
]

verkaufsdaten = pd.DataFrame(rohdaten)
verkaufsdaten["Gesamtsumme"] = verkaufsdaten["Menge"] * verkaufsdaten["Einzelpreis"]
print(verkaufsdaten)
Filiale = verkaufsdaten.groupby(by="Filiale", as_index=False).agg(
    Gesamtumsatz=("Gesamtsumme", "sum"),
    DurchschnittsBon=("Gesamtsumme", "mean")
)
print(f"\n {Filiale}")

Top_Filiale = Filiale[Filiale["Gesamtumsatz"] > 700]
print(f"\n {Top_Filiale}")

Top_Filiale_sort = Top_Filiale.sort_values(by="Gesamtumsatz", ascending=False)
print(f"\n {Top_Filiale_sort}")

Top_Filiale_sort.to_csv("filialen_bericht.csv", index=False)
Top_Filiale_sort.to_excel("filalen_bericht.xlsx", index=False)
```
