### Задача: «Mitarbeiter-Bonus-Analyse» (Анализ премий сотрудников)

Тебе нужно написать класс **`BonusAnalyse`**, который обрабатывает данные о продажах сотрудников и рассчитывает их итоговые бонусы.

#### Что должен уметь твой класс:

1. **Конструктор `__init__(self, daten)**`:
* Принимает список словарей с данными о совершенных сделках (`Mitarbeiter`, `Abteilung`, `Umsatz`).


2. **Метод `verarbeiten(self)**`:
* Превращает входящие данные в `pd.DataFrame`.
* Группирует данные по сотрудникам (`Mitarbeiter`).
* Для каждого сотрудника рассчитывает:
* **`Gesamtumsatz`** — суммарные продажи (сумма по полю `Umsatz`).
* **`AnzahlDeals`** — количество сделок (количество записей / `"count"`).


* Отфильтровывает только тех сотрудников, у которых **`Gesamtumsatz` больше 5000 €**.
* Добавляет новый столбец **`Bonus`**, который равен **10%** от их `Gesamtumsatz` (умножить `Gesamtumsatz` на `0.10`).
* Сортирует итоговую таблицу по столбцу `Bonus` по убыванию.
* Сохраняет результат в файлы:
* `bonus_bericht.csv` (без индекса строк)
* `bonus_bericht.xlsx` (без индекса строк)


* Возвращает полученный DataFrame через `return`.



---

### Исходные данные для теста:

Скопируй этот список в начало скрипта для проверки работы своего класса:

```
import pandas as pd

rohdaten_mitarbeiter = [
    {"Mitarbeiter": "Anna", "Abteilung": "Vertrieb", "Umsatz": 3000.0},
    {"Mitarbeiter": "Stefan", "Abteilung": "IT", "Umsatz": 1500.0},
    {"Mitarbeiter": "Anna", "Abteilung": "Vertrieb", "Umsatz": 4000.0},
    {"Mitarbeiter": "Markus", "Abteilung": "Vertrieb", "Umsatz": 2000.0},
    {"Mitarbeiter": "Stefan", "Abteilung": "IT", "Umsatz": 4500.0},
    {"Mitarbeiter": "Markus", "Abteilung": "Vertrieb", "Umsatz": 1000.0}
]

class BonusAnalyse:
    def __init__(self, daten):
        self.daten = daten
    def Mitarbeiter_daten(self):
        df = pd.DataFrame(self.daten)
        df = df.groupby("Mitarbeiter", as_index = False).agg(
            Gesamtumsatz=("Umsatz", "sum"),
            AnzahlDeals=("Mitarbeiter", "count")
        )
        df = df[df["Gesamtumsatz"] > 5000]
        df["Bonus"] = df["Gesamtumsatz"] * 0.10
        df = df.sort_values(by="Bonus", ascending=False)
        df.to_csv("bonus_bericht.csv", index=False)
        df.to_excel("bonus_bericht.xlsx", index=False)

        return df

#anpassen die Namen
analyse = BonusAnalyse(rohdaten_mitarbeiter)
ergebniss = analyse.Mitarbeiter_daten()

print(ergebniss)
```
