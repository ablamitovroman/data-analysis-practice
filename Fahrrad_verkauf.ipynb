```python
import pandas as pd

# Наш исходный словарь с данными о продажах
daten = {
    "datum": ["2026-09-01", "2026-09-01", "2026-09-02", "2026-09-02", "2026-09-03"],
    "kategorie": ["Mountainbike", "Zubehör", "E-Bike", "Zubehör", "Citybike"],
    "menge": [1, 3, 1, 5, 2],
    "einzelpreis": [450.0, 15.0, 1200.0, 25.0, 280.0]
}

# 1. Создаем DataFrame
verkaeufe = pd.DataFrame(daten)

# Задача 1: Добавляем новый столбец gesamt (перемножение векторов)
verkaeufe["gesamt"] = verkaeufe["menge"] * verkaeufe["einzelpreis"]

# Задача 2: Фильтрация продаж >= 200 €
grosse_verkaeufe = verkaeufe[verkaeufe["gesamt"] >= 200]

# Задача 3: Подсчет общей выручки
gesamteinnahmen = verkaeufe["gesamt"].sum()

# Выводим результаты:
print("--- Исходная таблица с новым столбцом 'gesamt': ---")
print(verkaeufe)

print("\n--- Продажи от 200 € и выше: ---")
print(grosse_verkaeufe)

print(f"\nОбщая выручка (Gesamteinnahmen): {gesamteinnahmen:.2f} €")
```
